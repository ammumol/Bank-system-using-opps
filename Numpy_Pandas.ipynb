## **Numpy** and **Pandas**

######1. Create a numpy array containing the numbers from 1 to 10, and then reshape it to a 2x5 matrix

```python
import numpy as np
arr=np.arange(1,11)
print("The array containing numbers from 1 to 10: ",arr)
new_arr=arr.reshape(2,5)
print("\n2X5 matrix :")
new_arr
```

2. Create a numpy array containing the numbers from 1 to 20, and then extract the elements between the 5th and 15th index.

```python
a=np.arange(1,21)
print("The array is ",a)
print("5th to 15th index elements are ",a[5:15])
```

3. How to compute the mean, median, and standard deviation of a numpy array?Use the
array you created above

```python
print("mean = ",np.mean(a))
print("median = ",np.median(a))
print("standard deviation = ",np.std(a))
```

4. Write a NumPy program that creates a 2D array x of shape (3, 4) and a 1D array y of
shape (4,). Subtract y from each row of x using broadcasting.

```python
x=np.array([[1,3,4,6],[12,8,6,10],[6,20,9,7]])
y=np.array([4,7,25,12])
print("3X4 array: \n",x)
print("1D array",y)
result=x-y
print("X-Y=\n",result)
```
